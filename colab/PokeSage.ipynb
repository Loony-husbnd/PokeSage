Showdown 빌드는 처음 한 번만 수 분 걸립니다.

```python
import subprocess, re
def sh(cmd):
    r = subprocess.run(cmd, shell=True, capture_output=True, text=True); return (r.stdout + r.stderr).strip()
v = sh("node --version")
print("node:", v or "없음")
m = re.match(r"v(\d+)", v)
if not m or int(m.group(1)) < 18:
    print(sh("curl -fsSL https://deb.nodesource.com/setup_20.x | bash - && apt-get install -y nodejs | tail -3"))
    print("node:", sh("node --version"))
```
